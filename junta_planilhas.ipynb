Notebook para juntar csvs com as informações coletadas e estimar dados importantes

```python
import pandas as pd
import re
```

```python
data_list = []
padrao_avaliadores = 'avaliador\d$'
for i in range(2006, 2025):
    data = pd.read_csv(f"csvs/final_defesas{i}.csv", index_col=0)
    avaliadores = [col for col in data.columns if re.search(padrao_avaliadores, col)]
    data['ano'] = i
    data_list.append(data)
```

```python
data_final = pd.concat(data_list)
```

```python
data_final.head()
```

```python
print(f"Total de defesas antes de limpeza = {data_final.shape[0]}")
```

```python
data_final = data_final[(data_final["orientador"] == data_final["orientador"]) & (data_final["orientador"] != "Selenium_error")]
```

```python
data_final["ano"].value_counts()
```

```python
for ano in data_final["ano"].value_counts(sort=False).index:
    print(f"{ano}:")
    data_final_ano = data_final[data_final["ano"] == ano]
    for curso, qtd in zip(data_final_ano["tipo"].value_counts().index, data_final_ano["tipo"].value_counts().values):
        print(f"    {curso}: {qtd}")
```

## Adicionar informações de gênero

```python
gender_data = pd.read_csv("csvs/genero.csv")
```

```python
gender_data.head()
```

```python
avaliadores = [col for col in data_final.columns if re.search(padrao_avaliadores, col)]
avaliadores.append("orientador")
```

```python
avaliadores
```

```python
for col in avaliadores:
    data_final[f"{col}_genero"] = pd.NA
```

```python
from collections import defaultdict
```

```python
generos = defaultdict(list)
for col in avaliadores:
    for orcid, idx in zip(data_final[col].values, data_final[col].index):
        if orcid == "sem orcid": 
            # generos[col].append((orcid, idx, "sem orcid", ""))
            generos[col].append({
                "orcid": orcid, "idx": idx, "gênero": "sem orcid", "nome orcid": 'sem orcid'
            })
            continue
        if gender_data[gender_data["orcid"] == orcid].shape[0] == 0: # não tem gênero ou nome da pessoa anotado para esse orcid
            generos[col].append({
                "orcid": orcid, "idx": idx, "gênero": "não anotado", "nome orcid": 'não anotado'
            })
            continue
        genero = gender_data[gender_data["orcid"] == orcid]["genero"].values[0]
        nome = gender_data[gender_data["orcid"] == orcid]["nome"].values[0]
        generos[col].append({
                "orcid": orcid, "idx": idx, "gênero": genero, "nome orcid": nome
            })
```

```python
for col, infos in generos.items():
    for info in infos:
        idx = info["idx"]
        genero = info["gênero"]
        data_final.at[idx, f"{col}_genero"] = genero
```
