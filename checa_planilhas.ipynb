# Verificação de quais planilhas tiveram erros de navegação no Selenium

```python
padrao_avaliadores = 'avaliador\d$'
for i in range(2006, 2025):
    data = pd.read_csv(f"csvs/final_defesas{i}.csv", index_col=0)
    avaliadores = [col for col in data.columns if re.search(padrao_avaliadores, col)]
    data_avaliadores = data[avaliadores + ["orientador"]]
    for col in data_avaliadores.columns:
        # if "Selenium_error" in data_avaliadores[col].values:
        #     print(i)
        for idx in data_avaliadores[col].index:
            if data_avaliadores.at[idx, col] == "Selenium_error":
                print(i, data.at[idx, "url"])
    # break
```
